```python
def generate_prompt(topic, task):
    prompts = {
        "Explain": f"""
Explain the topic "{topic}" in very simple language.

Include:
1. Definition
2. Important concepts
3. Simple example
4. Real-world application
5. Short conclusion

Make it suitable for a college student.
""",

        "Summary": f"""
Give a simple and clear summary of "{topic}".

Include:
- Main idea
- Important points
- Key terms
- Important facts

Keep the explanation easy to understand.
""",

        "Quiz": f"""
Create a quiz about "{topic}".

Create:
- 5 multiple-choice questions
- 4 options for each question
- Correct answer
- Short explanation for each answer

Make the questions suitable for a college student.
""",

        "Study Tips": f"""
Give effective study tips for learning "{topic}".

Include:
- What to study first
- Important concepts
- How to remember the concepts
- Practice methods
- Common mistakes to avoid
""",

        "Study Material": f"""
Create complete study material for "{topic}".

Include:
1. Introduction
2. Definition
3. Important concepts
4. Detailed explanation
5. Examples
6. Applications
7. Advantages
8. Limitations
9. Practice questions
10. Conclusion

Use simple language.
""",

        "Study Plan": f"""
Create a 7-day study plan for learning "{topic}".

For each day include:
- Topics to study
- Recommended activity
- Practice task
- Revision task

Keep the plan realistic for a college student.
"""
    }

    return prompts[task]
```

```python
topic = input("Enter the topic you want to study: ")

prompt = generate_prompt(topic, "Explain")

print("\nGenerated AI Prompt:\n")
print(prompt)
```

```python
def study_assistant():
    print("=" * 50)
    print("       AI-POWERED STUDY ASSISTANT")
    print("=" * 50)

    topic = input("\nEnter the topic: ")

    while True:
        print("\nChoose an option:")
        print("1. Explain Topic")
        print("2. Generate Summary")
        print("3. Generate Quiz")
        print("4. Generate Study Tips")
        print("5. Generate Complete Study Material")
        print("6. Generate 7-Day Study Plan")
        print("7. Exit")

        choice = input("\nEnter your choice: ")

        options = {
            "1": "Explain",
            "2": "Summary",
            "3": "Quiz",
            "4": "Study Tips",
            "5": "Study Material",
            "6": "Study Plan"
        }

        if choice in options:
            task = options[choice]

            prompt = generate_prompt(topic, task)

            print("\n" + "=" * 50)
            print("COPY THE FOLLOWING PROMPT INTO AN AI TOOL")
            print("=" * 50)
            print(prompt)
            print("=" * 50)

        elif choice == "7":
            print("\nThank you for using AI Study Assistant!")
            break

        else:
            print("\nInvalid choice. Please select 1-7.")
```

```python
study_assistant()
```
